```python
# --------------------------------------------
# Z-SCORE CALCULATION EXAMPLE
# --------------------------------------------

import math

# Step 1: Create a small dataset manually
data = [10, 20, 30, 40, 50]
print("Data:", data)

# Step 2: Calculate Mean
# Mean = Sum of values / Total number of values
mean = sum(data) / len(data)
print("Mean:", mean)

# Step 3: Calculate Variance
# Variance = Σ (X - Mean)^2 / N
variance = sum((x - mean) ** 2 for x in data) / len(data)
print("Variance:", variance)

# Step 4: Calculate Standard Deviation
# Standard Deviation = √Variance
std_dev = math.sqrt(variance)
print("Standard Deviation:", std_dev)

# Step 5: Calculate Z-Score
# Z = (X - Mean) / Standard Deviation
z_scores = [(x - mean) / std_dev for x in data]
print("Z-Scores:", z_scores)

# --------------------------------------------
# Interpretation:
# Z = 0  -> value equals mean
# Z > 0  -> above mean
# Z < 0  -> below mean
# --------------------------------------------
```

```python
import random
import math

# Generate 5 random numbers
data = [random.randint(1, 50) for _ in range(5)]
print("Data:", data)

# Mean
mean = sum(data) / len(data)

# Variance
variance = sum((x - mean) ** 2 for x in data) / len(data)

# Standard deviation
std_dev = math.sqrt(variance)

# Z-score
z_scores = [(x - mean) / std_dev for x in data]

print("Z-Scores:", z_scores)
```

Standardization from Scratch

```python
# mean=np.mean(data)
# print(mean)

s,n=0,0
for i in x:
    s+=i
    n=n+1
mean = s/n
print(mean)
```

```python
# sdeviation =np.std(x)
# print(sdeviation)

sq=0
for i in x:
    sq=sq+((i-mean)**2)

std=(sq/len(x))**(1/2)
print(std)
```

```python
x_scaled=[]
for i in x:
    x_scaled.append((i-mean)/std)
print(x_scaled)
```
