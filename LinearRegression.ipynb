Create a function my_slope(a,b) which accept two points a,b as an input such as a=($x_i$,$y_i$) and b=($x_j$,$y_j$) where $x_i$,$y_i$,$y_j$,$x_j$ denotes the values in x y axis


```python
def my_slope(a, b):
    """Return the slope between two points a and b.
    Parameters
    - a: tuple-like (x_i, y_i)
    - b: tuple-like (x_j, y_j)
    """

    xi, yi = a
    xj, yj = b
    dx = xj - xi
    if dx == 0:
        raise ValueError("Slope undefined for vertical line (x_i == x_j)")
    return (yj - yi) / dx
```

```python
def my_slope(a,b):

    """"
    Docstring for my_slope
    :param a:coords of a 2d point 
    :param b:coords of a 2d point 
    output:slope 
    """ 
    dy=a[1]-b[1]
    dx=a[0]-b[0]
    return dy/dx
```
